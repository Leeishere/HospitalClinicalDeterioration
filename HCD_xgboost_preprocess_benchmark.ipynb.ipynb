# XGBoost Preprocessing Benchmark

This notebook runs the fixed preprocessing configuration requested for XGBoost:
- Numeric transformer: custom_transformer
- Categorical encoder: ordinalencoder
- Outlier clipping: disabled
- Balanced training: enabled

It also prints best-model hyperparameters and train/test/unbalanced metrics for Recall, Precision, F1, and ROC-AUC.


```python
import pandas as pd
from preprocessing_benchmark_framework import (
    print_best_model_metrics,
    run_preprocessing_benchmark,
)

run_randomized = True
run_explicit = False
USE_IMBALANCE_OPTIONS = False
BALANCE_CLASSES_BEFORE_TRAINING = True

NUM_CURRENT_NOTEBOOKS = 4
TOTAL_THREADS = 16
```

```python
results_df, strategy_artifacts, top_per_metric, top_scores_df, top_combined, _ = run_preprocessing_benchmark(
    model_type='xgboost',
    variant='default',
    output_dir_name='xgboost',
    run_randomized=run_randomized,
    run_explicit=run_explicit,
    n_random_iterations_override=2,
    use_imbalance_options=USE_IMBALANCE_OPTIONS,
    balance_classes_before_training=BALANCE_CLASSES_BEFORE_TRAINING,
    total_threads=TOTAL_THREADS,
    num_current_notebooks=NUM_CURRENT_NOTEBOOKS,
    top_n=25,
    metrics_to_save=['recall', 'precision', 'f1', 'roc_auc'],
)
```

```python
pd.set_option("display.max_columns", None)

priority = ['recall', 'precision', 'accuracy', 'f1', 'roc_auc']

print_best_model_metrics(results_df, metrics_to_show=priority)

print('Top score per metric (top-1 by metric):')
display(top_scores_df)

print('Combined top-N shortlist (deduplicated, sorted by ROC-AUC):')
display(top_combined)
```
